```python
import seaborn as sns
%matplotlib inline
sns.set(font_scale=1.5)
import pandas as pd
```

```python
data=sns.load_dataset("flights")\
        .pivot("month","year","passengers") 
data.head() 
```

```python
data=pd.DataFrame(data=[[1,2,3],[3,2,1],[2,3,1]],columns=['a','b','c'],index=['d','e','f'])
```

```python
sns.set_context({"figure.figsize":(8,8)})
sns.heatmap(data=data,square=True,annot=True,fmt='d',cmap='RdBu_r')
```
