```python
import pandas as pd
import numpy as np
df=pd.read_csv(r'E:\wangzhilin\QuantumGalaxy\server_new\blueprints\map\static\委员信息表地图用_履历_表格视图 1.csv',)
df.columns=['name','year','sheng','shi','qu','story','other']
df = df.replace({np.nan: None})
df.sheng.replace({None:''},inplace=True)
df.shi.replace({None:''},inplace=True)
df.qu.replace({None:''},inplace=True)
df['city']=df.sheng+df.shi+df.qu
df.drop(['sheng','shi','qu'],axis=1,inplace=True)
df.year=df.year.astype('int')
df['story_start_in']=df['year']
df=df[['name','year','city','story','story_start_in','other']]
#df=df.where(df.notnull(),None,)

df = df.replace({np.nan: None})

df
```

```python
years=df.year.sort_values().to_list()
start_year,end_year=years[0],years[-1]
yearslist=[]
for i in range(end_year-start_year+1):
    yearslist.append(start_year+i)
yearslist
new_df=pd.DataFrame(data=yearslist,columns=['year'])

```

```python
persondf=new_df.merge(persondf,on='year',how='outer')
persondf
```

```python
final_df=pd.DataFrame(columns=['name','year','city','story','story_start_in','other'])
for person,persondf in df.groupby('name'):
    persondf=new_df.merge(persondf,on='year',how='outer')
    
    persondf['name']=person
    persondf.sort_values('year',inplace=True)
    persondf['city'].fillna(method='pad',inplace=True)
    persondf['story'].fillna(method='pad',inplace=True)
    persondf['story_start_in'].fillna(method='pad',inplace=True)
    persondf.dropna(axis=0,subset='story',inplace=True)
    final_df=pd.concat([final_df,persondf],axis=0)
    


final_df.sort_values('year',inplace=True)
final_df['story_start_in']=final_df.story_start_in.astype('int')
```

```python

final_df
```

```python
dataset_p=[]
for year,yeardata in final_df.groupby('year'):
    data={'source':None}
    source=[]
    #oneyear
    #print(yeardata)
    yeardata=yeardata.sort_values('name')
    for i,r in yeardata.iterrows():
        #print(r)
        coord=get_address(r.city)
        if coord:
            lng,lat=coord
            source.append([lng,lat,r['name'],r.city,r.story,r.year,r.story_start_in])
    data['source']=source
    dataset_p.append(data)
    
dataset_p
```

```python
js=json.dumps(dataset_p)
f2 = open(r'E:\wangzhilin\QuantumGalaxy\server_new\blueprints\map\static\js.json', 'w')
f2.write(js)
f2.close()
```
