```python
def send_msg(content,receive_id='ou_e32afa3ec16eae9f334d27c02c037259',msg_type='text'):
#默认值为王之霖id，测试用
    url = "https://open.feishu.cn/open-apis/im/v1/messages"
    params = {"receive_id_type":"open_id"}
    msg = content
    msgContent = {
        "text": msg,
    }
    req = {
        "receive_id": receive_id, # chat id
        "content": json.dumps(msgContent),
        "msg_type": msg_type,
    }
    payload = json.dumps(req)
    headers = {
        'Authorization': 'Bearer '+ fetch_tenant_token(), # your access token
        'Content-Type': 'application/json'
    }
    response = requests.request("POST", url, params=params, headers=headers, data=payload)
    #print(response.headers['X-Tt-Logid']) # for
    # 
    #  debug or oncall
    return response # Print Response
r=send_msg(content='abc')
```
