```python
def read_file(fpath): 
    BLOCK_SIZE = 1024*4
    with open(fpath, 'rb') as f: 
        while True: 
            block = f.read(BLOCK_SIZE) 
            if block: 
                yield block 
            else: 
                return
```

```python
import time
def work(files):
    for name,file_token in files.items():
        logger.info('开始下载%s骨干的图片'%name)
        urls=get_urls(file_token)
        num=0
        if urls['code']==0:
            urls=urls['urls']
            for record_id,url in urls.items():
                result,msg=download(url,file_token,record_id)
                if result:
                    num+=1
                else:
                    logger.info("%s:%s"%(record_id,msg))
                time.sleep(0.2)
            logger.info('完成%s下载，共%s张有效图片\n'%(name,num))
        else:
            logger.info('%s下载失败：%s\n'%(name,urls['msg']))
        
```
