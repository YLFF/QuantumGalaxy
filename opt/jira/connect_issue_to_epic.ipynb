```python
import pandas as pd
from jira import JIRA
import sys
sys.path.append('E:\wangzhilin\QuantumGalaxy')
from QGI.mysql import get_mysql
mysql=get_mysql()
jira=JIRA('https://research.quantumgalaxy.cn/', basic_auth=('bot2', "jira_bot2"))
```

```python
#受世界线影响：.fields.customfield_11412
```

```python
df=pd.read_csv(r'E:\wangzhilin\QuantumGalaxy\世界线_标的_标的全表.csv')
df=df[['公司名称','公司简述','相关世界线']]
df['code']=None
df
```

```python
for i,r in df.iterrows():
    name=r['公司名称']
    sql='select code from ticker_info where name like "%%%s%%"'%name
    code=mysql.read_query(sql)
    if code:
        df.loc[i,'code']=code[0][0]
        pass
    else:
        print(name)
```

```python
df = df.fillna(value=0)
```

```python
df
```

```python
epics={}
'a' in epics.keys()
```

```python
groups={}
for i,r in df.iterrows():
     if r['相关世界线'] and r['code']:
            target_epics=r['相关世界线'].split(',')
            #print(target_epics)
            issues=jira.search_issues(f'project = COMPSTUDY AND issuetype = 公司研究 AND 万得代码 ={r["code"]}')
            if not issues:
                 print(f'r["code"] has no jira issue')
            else:
                 issue=issues[0]
                 current_epics=issue.fields.customfield_11412
                 epics=[]
                 if current_epics:
                      current_epics=current_epics.split(', ')
                      #list or None
                      
                      for e in current_epics:
                           name=jira.issue(e).fields.summary
                           epics.append(name)
                 for target_epic in target_epics:
                      if target_epic in epics: #jiraissue已经关联了此世界线
                           print(target_epic)
                           continue
                      else:  
                           if target_epic in groups.keys():
                                groups[target_epic].append((r['公司名称'],r['code']))
                                pass
                           else:
                                groups[target_epic]=[]
                                groups[target_epic].append((r['公司名称'],r['code']))
                           pass
                           #将此世界线的id添加到issue的世界线field
                      
                      

    

     else:
        #print(r['公司名称'])
        pass
                 
```
