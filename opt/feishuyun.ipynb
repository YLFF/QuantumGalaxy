```python
def fetch_fld_children(folder_token,my_token='u-1L1WjFXql4iEj60PabvLYkhglQIglkm1Ogw0glSw25Qh',page_token=None):
    url='https://open.feishu.cn/open-apis/drive/v1/files'
    headers={'Authorization':"Bearer "+my_token}
    params={'page_size':200,
        'folder_token':folder_token}
    if page_token:
        params['page_token']=page_token
    r=base_fetch_func(url,headers,params=params,method='get')
    return r
def fetch_fld_meta(token,my_token='u-1L1WjFXql4iEj60PabvLYkhglQIglkm1Ogw0glSw25Qh'):
    url='https://open.feishu.cn/open-apis/drive/explorer/v2/folder/'+token+'/meta'
    header={'Authorization':"Bearer "+my_token,
    'Content_type':"application/json; charset=utf-8"}
    r=base_fetch_func(url,header,method='get')
    return r
def move_file(target,type,token,my_token):
    url='https://open.feishu.cn/open-apis/drive/v1/files/'+token+'/move'
    headers={'Authorization':"Bearer "+my_token,
    'Content_type':"application/json; charset=utf-8"}
    data={'type':type,
    'folder_token':target}
    r=base_fetch_func(url,headers,data=data,method='post')
    return r
```
