```python
r=neo.fetch_company_info()
df=pd.DataFrame(r,columns=['neoname','neonickname','neocode'],index=None)
df

```

```python
df=df.replace({None:''}).sort_values('neocode',ascending=False).reindex()
df['与数据库代码一致']=False
df['qgdbscode'],df['qgdbsname']=None,None
```

```python
df1=df
df1
```

```python
for i,r in df1.iterrows():
    if r.neocode!='':
        sql='select name,code from ticker_info where code="%s"'%(r.neocode)
        result=mysql.read_query(sql)
        if result:
            df1.loc[i,'qgdbsname']=result[0][0]
            df1.loc[i,'qgdbscode']=result[0][1]
            df1.loc[i,'与数据库代码一致']=True
    else:
        if r.neonickname !='':
            print(r.neoname,r.neonickname)
            sql='select name,code from ticker_info where name like "%%%s%%" or name like "%%%s%%" limit 1'%(r.neoname,r.neonickname)
        else:
            print(r.neoname,'none')
            sql='select name,code from ticker_info where name like "%%%s%%"  limit 1'%(r.neoname)
        result=mysql.read_query(sql)
        print(result)
        if result:
            df1.loc[i,'qgdbsname']=result[0][0]
            df1.loc[i,'qgdbscode']=result[0][1]
df1
```

```python
df1.sort_values(by=['与数据库代码一致','qgdbscode'],ascending=False).to_excel('cache.xlsx')
```
